#### 1. Phân biệt toán tử định dạng chuỗi và hàm định dạng chuỗi có sẵn trong Python? Cho năm ví dụ minh họa.
Toán tử định dạng chuỗi (%): Dùng ký hiệu % để chèn giá trị vào chuỗi.
Hàm định dạng chuỗi (format() hoặc f-string): Cú pháp hiện đại, dễ đọc, hỗ trợ nhiều kiểu định dạng.

```python
name = "Hoàng"
age = 20
score = 8.5

print("Tên: %s" % name)
print("Tuổi: %d" % age)
print("Điểm: %.1f" % score)
print("Tên: {}, Tuổi: {}".format(name, age))
print(f"Tên: {name}, Tuổi: {age}, Điểm: {score}")
```

#### 2. Viết chương trình xuất ra số ngẫu nhiên trong một đoạn bất kỳ cho trước?

```python
class NhanVien:
    def __init__(self, maNV: int, **kwargs):
        self.maNV = maNV
        self.hoTen = kwargs.get('hoTen')
        self.luongCB = kwargs.get('luongCB')

    # Getter
    def getMaNV(self):
        return self.maNV

    def getHoTen(self):
        return self.hoTen

    def getLuongCB(self):
        return self.luongCB

    # Setter
    def setMaNV(self, maNV):
        self.maNV = maNV

    def setHoTen(self, hoTen):
        self.hoTen = hoTen

    def setLuongCB(self, luongCB):
        self.luongCB = luongCB


nv = NhanVien(
    1,
    hoTen="Phan Huy Hoang",
    luongCB=20000000
)

print(nv.__dict__)

nv.setHoTen("phh235")
nv.setLuongCB(15000000)

print(nv.getMaNV())
print(nv.getHoTen())
print(nv.getLuongCB())

print(nv.__dict__)
```

```python
class PhuongTrinhBacNhat:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def giai(self):
        if self.a == 0:
            if self.b == 0:
                print("Phương trình vô số nghiệm")
            else:
                print("Phương trình vô nghiệm")
        else:
            x = -self.b / self.a
            print("Nghiệm x =", x)


pt = PhuongTrinhBacNhat(2, 4)
pt.giai()
```

```python
class PhuongTrinhBacHai:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def giai(self):
        if self.a == 0:
            if self.b == 0:
                if self.c == 0:
                    print("Phương trình vô số nghiệm")
                else:
                    print("Phương trình vô nghiệm")
            else:
                x = -self.c / self.b
                print("Nghiệm x =", x)
            return

        delta = self.b ** 2 - 4 * self.a * self.c

        if delta < 0:
            print("Phương trình vô nghiệm")

        elif delta == 0:
            x = -self.b / (2 * self.a)
            print("Phương trình có nghiệm kép x =", x)

        else:
            x1 = (-self.b + math.sqrt(delta)) / (2 * self.a)
            x2 = (-self.b - math.sqrt(delta)) / (2 * self.a)

            print("x1 =", x1)
            print("x2 =", x2)


pt = PhuongTrinhBacHai(1, -3, 2)
pt.giai()
```

```python
class PhuongTrinhTrungPhuong:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def giai(self):
        if self.a == 0:
            print("Không phải phương trình trùng phương")
            return

        delta = self.b ** 2 - 4 * self.a * self.c

        if delta < 0:
            print("Phương trình vô nghiệm")
            return

        if delta == 0:
            t = -self.b / (2 * self.a)

            if t < 0:
                print("Phương trình vô nghiệm")
            elif t == 0:
                print("x = 0")
            else:
                print("x1 =", math.sqrt(t))
                print("x2 =", -math.sqrt(t))

        else:
            t1 = (-self.b + math.sqrt(delta)) / (2 * self.a)
            t2 = (-self.b - math.sqrt(delta)) / (2 * self.a)

            nghiem = []

            if t1 >= 0:
                nghiem.append(math.sqrt(t1))
                if t1 != 0:
                    nghiem.append(-math.sqrt(t1))

            if t2 >= 0:
                nghiem.append(math.sqrt(t2))
                if t2 != 0:
                    nghiem.append(-math.sqrt(t2))

            if len(nghiem) == 0:
                print("Phương trình vô nghiệm")
            else:
                print("Các nghiệm:", nghiem)


pt = PhuongTrinhTrungPhuong(1, -5, 4)
pt.giai()
```

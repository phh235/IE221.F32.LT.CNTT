```python
import math
```

```python
# Tính tổng và tích nhiều số
def tinh_tong_va_tich(*args):
    tich = 1

    for so in args:
        tich *= so

    return sum(args), tich


tong, tich = tinh_tong_va_tich(2, 3, 4, 5)

print("Tổng:", tong)
print("Tích:", tich)
```

```python
# Tính chu vi hình vuông, chữ nhật, tròn

def tinh_chu_vi(hinh, **kwargs):
    if hinh == "vuong":
        return kwargs["canh"] * 4

    if hinh == "chu_nhat":
        return (kwargs["dai"] + kwargs["rong"]) * 2

    if hinh == "tron":
        return 2 * math.pi * kwargs["ban_kinh"]


print(tinh_chu_vi("vuong", canh=5))
print(tinh_chu_vi("chu_nhat", dai=10, rong=5))
print(tinh_chu_vi("tron", ban_kinh=3))
```

```python
# Tính chu vi hoặc diện tích

def tinh_chu_vi_dien_tich(hinh, *args, **kwargs):
    loai = args[0]

    if hinh == "vuong":
        canh = kwargs["canh"]
        return canh * 4 if loai == "chu_vi" else canh ** 2

    if hinh == "chu_nhat":
        dai = kwargs["dai"]
        rong = kwargs["rong"]
        return (dai + rong) * 2 if loai == "chu_vi" else dai * rong

    if hinh == "tron":
        r = kwargs["ban_kinh"]
        return 2 * math.pi * r if loai == "chu_vi" else math.pi * r ** 2


print(tinh_chu_vi_dien_tich("vuong", "chu_vi", canh=5))
print(tinh_chu_vi_dien_tich("vuong", "dien_tich", canh=5))

print(tinh_chu_vi_dien_tich("chu_nhat", "chu_vi", dai=10, rong=5))
print(tinh_chu_vi_dien_tich("chu_nhat", "dien_tich", dai=10, rong=5))

print(tinh_chu_vi_dien_tich("tron", "chu_vi", ban_kinh=3))
print(tinh_chu_vi_dien_tich("tron", "dien_tich", ban_kinh=3))
```

```python
# Tính căn bậc n của một số thực
def can_bac_n(x, n):
    return x ** (1 / n)


print(can_bac_n(16, 2))
```

```python
# Tính bình phương của một số dương
def binh_phuong(x):
    if x > 0:
        return x ** 2
    return "Phải nhập số dương"


print(binh_phuong(5))
```

```python
# Kiểm tra số chẵn và âm
def kiem_tra_chan_am(x):
    if x < 0 and x % 2 == 0:
        return True
    return False


print(kiem_tra_chan_am(-4))
```

```python
# Số âm lẻ => -1
# Số dương chẵn => 1
# Còn lại => 0

def kiem_tra_so(x):
    if x < 0 and x % 2 != 0:
        return -1

    if x > 0 and x % 2 == 0:
        return 1

    return 0


print(kiem_tra_so(-3))
print(kiem_tra_so(4))
print(kiem_tra_so(5))
```

```python
# Nhập số thuộc đoạn [-89, 90]

def nhap_so():
    x = int(input("Nhập số: "))

    while x < -89 or x > 90:
        print("Sai, nhập lại")
        x = int(input("Nhập số: "))

    return x


print(nhap_so())
```

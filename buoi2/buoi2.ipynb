```python
# Chuỗi ban đầu
s = "Đại học Quốc gia, Khu phố 6, P. Linh Trung, Q. Thủ Đức, Tp. HCM"
# Câu 1:
result1 = s.split(", ")

for item in result1:
    print(item)
```

```python
# Câu 2:
result2 = s.split(", ")

result2[2] = result2[2].replace("P. ", "")
result2[3] = result2[3].replace("Q. ", "")
result2[4] = result2[4].replace("Tp. ", "")

for item in result2:
    print(item)
```

```python
# 1. Index các phần tử đầu, cuối và giữa list
# Khởi tạo list
list = ["Python", 20, "CNTT", 3.14, 100]

print("List:", list)

print("Phần tử đầu:", list[0])
print("Phần tử cuối:", list[-1])

mid = len(list) // 2
print("Phần tử giữa:", list[mid])
```

```python
# 2. Slice từ vị trí begin đến end, begin và end nhập từ bàn phím.
begin = int(input("Nhập begin: "))
end = int(input("Nhập end: "))

print("Kết quả slice:")
print(list[begin:end])
```

```python
# 3. Thao tác thêm, xóa, sửa dữ liệu tại vị trí bất kỳ.
# Thêm
list.insert(2, "Java")
print("Sau khi thêm:")
print(list)

# Xóa
del list[3]
print("Sau khi xóa:")
print(list)

# Sửa
list[1] = 999
print("Sau khi sửa:")
print(list)
```

```python
set1 = {10, 15, 20, 25, 30, 35, 40, 45, 50, 55}
set2 = {20, 25, 30, 35, 40}

print("Set 1:", set1)
print("Set 2:", set2)
```

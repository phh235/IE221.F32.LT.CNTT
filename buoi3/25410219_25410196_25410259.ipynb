### Bài 1: Thống kê ký tự trong chuỗi

```python
str_input = input("Nhập chuỗi: ")

ky_tu_dac_biet = "!@#$%^&*()-=+./"

dem_dac_biet = 0
dem_chu_thuong = 0
dem_chu_so = 0
dem_chu_hoa = 0

chuoi_dac_biet = ""
chuoi_chu_thuong = ""
chuoi_chu_so = ""
chuoi_chu_hoa = ""

for ky_tu in str_input:
    if ky_tu in ky_tu_dac_biet:
        dem_dac_biet = dem_dac_biet + 1
        chuoi_dac_biet = chuoi_dac_biet + ky_tu + " "

    elif ky_tu >= "a" and ky_tu <= "z":
        dem_chu_thuong = dem_chu_thuong + 1
        chuoi_chu_thuong = chuoi_chu_thuong + ky_tu + " "

    elif ky_tu >= "0" and ky_tu <= "9":
        dem_chu_so = dem_chu_so + 1
        chuoi_chu_so = chuoi_chu_so + ky_tu + " "

    elif ky_tu >= "A" and ky_tu <= "Z":
        dem_chu_hoa = dem_chu_hoa + 1
        chuoi_chu_hoa = chuoi_chu_hoa + ky_tu + " "

print("\nKẾT QUẢ")
print("Độ dài chuỗi:", len(str_input))

print("Các ký tự đặc biệt:", chuoi_dac_biet)
print("Số lượng ký tự đặc biệt:", dem_dac_biet)

print("Các chữ cái thường [a-z]:", chuoi_chu_thuong)
print("Số lượng chữ thường:", dem_chu_thuong)

print("Các chữ số [0-9]:", chuoi_chu_so)
print("Số lượng chữ số:", dem_chu_so)

print("Các chữ cái hoa [A-Z]:", chuoi_chu_hoa)
print("Số lượng chữ hoa:", dem_chu_hoa)
```

### Bài 2: Viết hàm kiểm tra để phát hiện chuỗi nhập vào có phải là một e-mail hay không?

```python
def kiem_tra_email(email):
    danh_sach_ten_mien = [
        "gmail.com",
        "yahoo.com",
        "yahoo.com.vn",
        "hotmail.com",
        "outlook.com",
        "live.com",
        "icloud.com"
    ]

    # Kiểm tra khoảng trắng
    for ky_tu in email:
        if ky_tu == " ":
            return False

    # Đếm ký tự @
    dem_a_cong = 0

    for ky_tu in email:
        if ky_tu == "@":
            dem_a_cong = dem_a_cong + 1

    if dem_a_cong != 1:
        return False

    # Tách phần trước và sau @
    vi_tri_a_cong = email.find("@")

    ten_nguoi_dung = email[0:vi_tri_a_cong]
    ten_mien = email[vi_tri_a_cong + 1:len(email)]

    # Phần trước @ phải từ 6 ký tự
    if len(ten_nguoi_dung) < 6:
        return False

    # Phần trước @ chỉ được chứa chữ và số
    for ky_tu in ten_nguoi_dung:
        la_chu_thuong = ky_tu >= "a" and ky_tu <= "z"
        la_chu_hoa = ky_tu >= "A" and ky_tu <= "Z"
        la_chu_so = ky_tu >= "0" and ky_tu <= "9"

        if not la_chu_thuong and not la_chu_hoa and not la_chu_so:
            return False

    # Kiểm tra tên miền
    ten_mien_hop_le = False

    for mien in danh_sach_ten_mien:
        if ten_mien == mien:
            ten_mien_hop_le = True

    if ten_mien_hop_le == False:
        return False

    return True


email = input("Nhập email: ")

if kiem_tra_email(email):
    print("Email hợp lệ.")
else:
    print("Email không hợp lệ.")
```

### Bài 3: Kiểm tra ID User và Password

```python
def kiem_tra_user_id(user_id):
    ky_tu_cam = "!@#$%^&*()-=+"

    if len(user_id) < 6:
        print("ID User phải có ít nhất 6 ký tự.")
        return False

    if len(user_id) > 24:
        print("ID User chỉ được có tối đa 24 ký tự.")
        return False

    for ky_tu in user_id:
        if ky_tu == " ":
            print("ID User không được chứa khoảng trắng.")
            return False

        if ky_tu in ky_tu_cam:
            print("ID User chứa ký tự không hợp lệ:", ky_tu)
            return False

    return True


def kiem_tra_password(password):
    if len(password) < 6:
        print("Password phải có ít nhất 6 ký tự.")
        return False

    if len(password) > 24:
        print("Password chỉ được có tối đa 24 ký tự.")
        return False

    co_chu_thuong = False
    co_chu_hoa = False
    co_chu_so = False
    co_ky_tu_dac_biet = False

    for ky_tu in password:
        if ky_tu >= "a" and ky_tu <= "z":
            co_chu_thuong = True

        elif ky_tu >= "A" and ky_tu <= "Z":
            co_chu_hoa = True

        elif ky_tu >= "0" and ky_tu <= "9":
            co_chu_so = True

        elif ky_tu == "$" or ky_tu == "#" or ky_tu == "@":
            co_ky_tu_dac_biet = True

    if co_chu_thuong == False:
        print("Password phải có ít nhất một chữ thường.")
        return False

    if co_chu_hoa == False:
        print("Password phải có ít nhất một chữ hoa.")
        return False

    if co_chu_so == False:
        print("Password phải có ít nhất một chữ số.")
        return False

    if co_ky_tu_dac_biet == False:
        print("Password phải có ít nhất một ký tự $, # hoặc @.")
        return False

    return True


user_id = input("Nhập ID User: ")
password = input("Nhập Password: ")

user_hop_le = kiem_tra_user_id(user_id)
password_hop_le = kiem_tra_password(password)

if user_hop_le == True and password_hop_le == True:
    print("ID User và Password hợp lệ.")
    print("Đăng nhập thành công.")
else:
    print("Đăng nhập thất bại.")
```

### Bài 4: Kéo – Búa – Bao giữa người và máy

### Bài 5: Nhiều người tự động chơi Kéo – Búa – Bao

```python
import random


def tim_nguoi_thang(lua_chon_1, lua_chon_2):
    if lua_chon_1 == lua_chon_2:
        return 0

    elif lua_chon_1 == 1 and lua_chon_2 == 3:
        return 1

    elif lua_chon_1 == 2 and lua_chon_2 == 1:
        return 1

    elif lua_chon_1 == 3 and lua_chon_2 == 2:
        return 1

    else:
        return 2


def in_lua_chon(lua_chon):
    if lua_chon == 1:
        return "Kéo"

    elif lua_chon == 2:
        return "Búa"

    else:
        return "Bao"


so_nguoi = random.randrange(8, 21)

danh_sach_nguoi = []

for i in range(1, so_nguoi + 1):
    danh_sach_nguoi.append(i)

print("Số người chơi:", so_nguoi)

vong = 1

while len(danh_sach_nguoi) > 1:
    print("\n===== VÒNG", vong, "=====")

    danh_sach_vong_sau = []
    vi_tri = 0

    while vi_tri < len(danh_sach_nguoi):
        # Trường hợp còn một người không có đối thủ
        if vi_tri == len(danh_sach_nguoi) - 1:
            nguoi = danh_sach_nguoi[vi_tri]

            print("Người", nguoi, "không có đối thủ.")
            print("Người", nguoi, "được vào vòng sau.")

            danh_sach_vong_sau.append(nguoi)
            break

        nguoi_1 = danh_sach_nguoi[vi_tri]
        nguoi_2 = danh_sach_nguoi[vi_tri + 1]

        da_co_nguoi_thang = False

        while da_co_nguoi_thang == False:
            lua_chon_1 = random.randrange(1, 4)
            lua_chon_2 = random.randrange(1, 4)

            print(
                "Người", nguoi_1, "chọn", in_lua_chon(lua_chon_1),
                "- Người", nguoi_2, "chọn", in_lua_chon(lua_chon_2)
            )

            ket_qua = tim_nguoi_thang(lua_chon_1, lua_chon_2)

            if ket_qua == 0:
                print("Hai người hòa, chơi lại.")

            elif ket_qua == 1:
                print("Người", nguoi_1, "thắng.")
                danh_sach_vong_sau.append(nguoi_1)
                da_co_nguoi_thang = True

            else:
                print("Người", nguoi_2, "thắng.")
                danh_sach_vong_sau.append(nguoi_2)
                da_co_nguoi_thang = True

        vi_tri = vi_tri + 2

    danh_sach_nguoi = danh_sach_vong_sau
    vong = vong + 1

print("\nNGƯỜI VÔ ĐỊCH LÀ NGƯỜI", danh_sach_nguoi[0])
```

### Bài 6: Mô phỏng vé số Vietlott 6/45

```python
def nhap_mot_ve(so_thu_tu):
    danh_sach_so = []

    print("\nNhập vé số thứ", so_thu_tu)

    while len(danh_sach_so) < 6:
        try:
            so = int(input("Nhập một số từ 1 đến 45: "))
        except ValueError:
            print("Vui lòng nhập một số nguyên từ 1 đến 45.")
            continue

        if so < 1 or so > 45:
            print("Số phải nằm trong khoảng từ 1 đến 45.")

        elif so in danh_sach_so:
            print("Số này đã tồn tại trong vé.")

        else:
            danh_sach_so.append(so)

    return danh_sach_so


def tao_day_so_trung_thuong():
    day_so = []

    while len(day_so) < 6:
        so = random.randrange(1, 46)

        if so not in day_so:
            day_so.append(so)

    return day_so


def dem_so_trung(ve_so, day_so_trung_thuong):
    dem = 0

    for so in ve_so:
        if so in day_so_trung_thuong:
            dem = dem + 1

    return dem


def tinh_tien_thuong(so_luong_trung):
    if so_luong_trung == 3:
        return 30000

    elif so_luong_trung == 4:
        return 300000

    elif so_luong_trung == 5:
        return 10000000

    elif so_luong_trung == 6:
        return 10000000000

    else:
        return 0


print("MÔ PHỎNG VÉ SỐ VIETLOTT 6/45")

so_ve = int(input("Nhập số lượng vé muốn mua: "))

while so_ve <= 0:
    print("Số lượng vé phải lớn hơn 0.")
    so_ve = int(input("Nhập lại số lượng vé: "))

danh_sach_ve = []

for i in range(1, so_ve + 1):
    ve = nhap_mot_ve(i)
    danh_sach_ve.append(ve)

day_so_trung_thuong = tao_day_so_trung_thuong()

print("\nDãy số trúng thưởng:", day_so_trung_thuong)

tong_tien_thuong = 0

for i in range(0, len(danh_sach_ve)):
    ve = danh_sach_ve[i]

    so_luong_trung = dem_so_trung(
        ve,
        day_so_trung_thuong
    )

    tien_thuong = tinh_tien_thuong(so_luong_trung)

    tong_tien_thuong = tong_tien_thuong + tien_thuong

    print("\nVé số thứ", i + 1, ":", ve)
    print("Số lượng số trùng:", so_luong_trung)
    print("Tiền thưởng:", tien_thuong, "đồng")

tong_tien_mua_ve = so_ve * 10000

print("\n===== THỐNG KÊ =====")
print("Số vé đã mua:", so_ve)
print("Tổng tiền mua vé:", tong_tien_mua_ve, "đồng")
print("Tổng tiền thưởng:", tong_tien_thuong, "đồng")

if tong_tien_thuong > tong_tien_mua_ve:
    tien_lai = tong_tien_thuong - tong_tien_mua_ve
    print("Người chơi lãi:", tien_lai, "đồng")

elif tong_tien_thuong < tong_tien_mua_ve:
    tien_lo = tong_tien_mua_ve - tong_tien_thuong
    print("Người chơi lỗ:", tien_lo, "đồng")

else:
    print("Người chơi hòa vốn.")
```

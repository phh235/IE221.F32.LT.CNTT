```python
import random

def nhap_ve():
    ve = []

    while len(ve) < 6:
        so = int(input(f"Số thứ {len(ve) + 1}: "))

        if 1 <= so <= 45 and so not in ve:
            ve.append(so)
        else:
            print("Số không hợp lệ hoặc bị trùng")

    return ve

def xo_so():
    ket_qua = []

    while len(ket_qua) < 6:
        so = random.randint(1, 45)

        if so not in ket_qua:
            ket_qua.append(so)

    return ket_qua

def dem_so_trung(ve, ket_qua):
    dem = 0

    for so in ve:
        if so in ket_qua:
            dem += 1

    return dem

def tinh_tien_thuong(so_trung):
    if so_trung == 3:
        return 30000
    elif so_trung == 4:
        return 300000
    elif so_trung == 5:
        return 10000000
    elif so_trung == 6:
        return 10000000000

    return 0

so_luong_ve = int(input("Nhập số lượng vé: "))
danh_sach_ve = []

for i in range(so_luong_ve):
    print(f"\n--- Vé {i + 1} ---")
    danh_sach_ve.append(nhap_ve())

ket_qua = xo_so()
print("\nDãy số trúng thưởng:", ket_qua)

tong_tien = 0

for i in range(len(danh_sach_ve)):
    ve = danh_sach_ve[i]
    so_trung = dem_so_trung(ve, ket_qua)
    tien_thuong = tinh_tien_thuong(so_trung)

    tong_tien += tien_thuong

    print(f"\nVé {i + 1}: {ve}")
    print("Số trùng:", so_trung)
    print("Tiền thưởng:", tien_thuong, "đ")

print("\n--- THỐNG KÊ ---")
print("Tiền mua vé:", so_luong_ve * 10000, "đ")
print("Tổng tiền thưởng:", tong_tien, "đ")
```
